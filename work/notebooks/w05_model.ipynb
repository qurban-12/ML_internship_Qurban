## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

I will use a grouped train/test split by `client_id`. This keeps all content from the same client in only one side of the split, which reduces the risk of learning client-specific patterns.

I will use 80% of the data for training and 20% for testing, with a fixed random seed for reproducibility. The same test split will be used to evaluate both the Week-4 baseline and the new model.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.model_selection import GroupShuffleSplit

# Grouped train/test split by client
X = df[MODEL_NUMERIC_FEATURES + MODEL_CATEGORICAL_FEATURES].copy()
y = df[TARGET].copy()
groups = df[GROUP_COLUMN].copy()

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

groups_train = groups.iloc[train_idx].copy()
groups_test = groups.iloc[test_idx].copy()

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining clients:", groups_train.nunique())
print("Testing clients:", groups_test.nunique())

print("\nClients shared between train and test:",
      len(set(groups_train) & set(groups_test)))

print("\nTraining target rate:", round(y_train.mean() * 100, 2), "%")
print("Testing target rate:", round(y_test.mean() * 100, 2), "%")
```

## 3. Train + compare vs my baseline

*Same data, same metric, same split as your Week-4 baseline. Show the table.*

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

# Preprocessing for numeric features
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Preprocessing for categorical features
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, MODEL_NUMERIC_FEATURES),
    ("categorical", categorical_pipeline, MODEL_CATEGORICAL_FEATURES)
])

# Logistic Regression model
logistic_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Train
logistic_model.fit(X_train, y_train)

# Predicted probability of declining
logistic_scores = logistic_model.predict_proba(X_test)[:, 1]

print("Logistic Regression trained successfully.")
print("Number of test predictions:", len(logistic_scores))
print("Minimum predicted score:", round(logistic_scores.min(), 4))
print("Maximum predicted score:", round(logistic_scores.max(), 4))
```

```python
# Recreate the Week-4 baseline rule

STALE_DAYS = 91
GOOD_POSITION_MAX = 10
LOW_CTR_MAX = 0.10

baseline_score_all = (
    (df["days_since_last_update"] >= STALE_DAYS).astype(int) * 2
    +
    (
        (df["avg_position"] >= 1)
        & (df["avg_position"] <= GOOD_POSITION_MAX)
        & (df["ctr"] < LOW_CTR_MAX)
    ).astype(int)
)

# Evaluate baseline on the exact same test rows
baseline_scores_test = baseline_score_all.iloc[test_idx].to_numpy()

print("Baseline scores on test set:")
print(pd.Series(baseline_scores_test).value_counts().sort_index())

print("\nTest baseline scores:", len(baseline_scores_test))
print("Test labels:", len(y_test))
```

I trained a Logistic Regression model using the official 26 features and evaluated it on the same client-grouped test set as the Week-4 baseline.

On the test set, the base rate of declining content was 51.1%. The Week-4 baseline and Logistic Regression both measured 0.700 Precision@20. Logistic Regression measured 0.720 Precision@50 versus 0.620 for the baseline, and 0.700 Precision@100 versus 0.550 for the baseline.

These results are directional decision-support measurements on this test split. They show that Logistic Regression provides stronger precision at K=50 and K=100, while the two approaches were equal at K=20.


```python
from scripts.ml_utils import precision_at_k

# Precision@K comparison
K_VALUES = [20, 50, 100]

comparison_rows = []

for k in K_VALUES:
    comparison_rows.append({
        "Method": "Week-4 Baseline",
        "Precision@K": f"Precision@{k}",
        "Value": precision_at_k(y_test, baseline_scores_test, k)
    })

    comparison_rows.append({
        "Method": "Logistic Regression",
        "Precision@K": f"Precision@{k}",
        "Value": precision_at_k(y_test, logistic_scores, k)
    })

comparison_df = pd.DataFrame(comparison_rows)

print("Test-set base rate:",
      round(y_test.mean() * 100, 2), "%")

print("\nComparison:")
display(
    comparison_df.style.format({"Value": "{:.3f}"})
)
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score

logistic_roc_auc = roc_auc_score(y_test, logistic_scores)
logistic_pr_auc = average_precision_score(y_test, logistic_scores)

print("Logistic Regression ROC-AUC:", round(logistic_roc_auc, 3))
print("Logistic Regression Average Precision:", round(logistic_pr_auc, 3))
```

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

The model is useful for ranking declining content, but its predictions are not perfect. I will inspect classification errors and feature importance to understand where the model is making mistakes and what signals it relies on.

I will use permutation importance as a diagnostic. A feature with higher importance means that changing that feature affected the model's measured performance more on the test set. This shows association with the model's predictions, not a causal relationship.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.metrics import confusion_matrix, classification_report

# Convert probabilities to 0/1 predictions using a 0.5 threshold
logistic_predictions = (logistic_scores >= 0.5).astype(int)

# Confusion matrix
cm = confusion_matrix(y_test, logistic_predictions)

print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        logistic_predictions,
        target_names=["Not declining", "Declining"],
        digits=3
    )
)
```

The Logistic Regression model correctly identified 2,281 of the 3,149 declining items, giving a recall of 0.724 for the declining class. However, it also predicted 1,675 non-declining items as declining, which are false positives.

Overall accuracy was 0.587, so the model should not be treated as a perfect classifier. For this task, ranking quality is more relevant than a fixed 0.5 classification threshold because the main goal is to identify useful items for review. I will therefore inspect concrete errors and feature importance before drawing conclusions about the model.


```python
# Build an error-analysis dataframe from the test set
error_df = df.iloc[test_idx].copy()

error_df["actual"] = y_test.to_numpy()
error_df["predicted"] = logistic_predictions
error_df["prediction_score"] = logistic_scores

# False positives: predicted declining, but actually not declining
false_positives = error_df[
    (error_df["actual"] == 0) &
    (error_df["predicted"] == 1)
].copy()

# False negatives: actually declining, but predicted not declining
false_negatives = error_df[
    (error_df["actual"] == 1) &
    (error_df["predicted"] == 0)
].copy()

print("False positives:", len(false_positives))
print("False negatives:", len(false_negatives))

print("\nExample false positives:")
display(
    false_positives[
        [
            "content_id",
            "content_type",
            "days_since_last_update",
            "avg_position",
            "ctr",
            "prediction_score",
            "actual",
            "predicted"
        ]
    ].head(3)
)

print("\nExample false negatives:")
display(
    false_negatives[
        [
            "content_id",
            "content_type",
            "days_since_last_update",
            "avg_position",
            "ctr",
            "prediction_score",
            "actual",
            "predicted"
        ]
    ].head(3)
)
```

The concrete errors show why the model should be used as decision support rather than as a definitive classifier. Some false positives have signals associated with decline, such as older content or very low CTR, but their observed label is not declining. For example, one false positive was 103 days since its last update with a CTR of 0.00%, while another had been updated only 13 days earlier.

The false negatives show the opposite problem. Some items were actually declining but received relatively low predicted probabilities. For example, one false negative was 104 days since its last update with a CTR of 0.00%, but the model assigned it a probability of only 0.210.

These examples suggest that individual signals do not fully determine the observed outcome. This is one reason to evaluate the model as a ranking system using Precision@K rather than relying only on a 0.5 classification threshold.


```python
from sklearn.inspection import permutation_importance

# Measure how much each original feature affects model performance
perm = permutation_importance(
    logistic_model,
    X_test,
    y_test,
    scoring="roc_auc",
    n_repeats=10,
    random_state=42,
    n_jobs=-1
)

importance_df = pd.DataFrame({
    "feature": X_test.columns,
    "importance_mean": perm.importances_mean,
    "importance_std": perm.importances_std
}).sort_values(
    "importance_mean",
    ascending=False
)

print("Top 10 features by permutation importance:")

display(
    importance_df.head(10).style.format({
        "importance_mean": "{:.4f}",
        "importance_std": "{:.4f}"
    })
)
```

Permutation importance shows that `log_impressions_90d` was the most influential feature for the Logistic Regression model, with a mean importance of 0.0803. It was followed by `log_clicks_90d` at 0.0581 and `content_age_days` at 0.0342.

Other useful signals included `impression_tier`, `avg_position`, and `log_sessions_90d`. These results suggest that the model relies strongly on recent traffic and visibility-related measurements, together with content age and search-position information.

The importance values describe how much the features contribute to the model's measured predictive performance. They should not be interpreted as causal effects. The results also provide a useful sanity check because the important features are related to content visibility, traffic, age, and position rather than the target-defining fields `trend_direction` or `trend_pct`.


```python
# Final feature leakage check
forbidden_features = {
    "trend_direction",
    "trend_pct",
    "is_declining_label",
    "client_id",
    "content_id"
}

used_features = set(
    MODEL_NUMERIC_FEATURES + MODEL_CATEGORICAL_FEATURES
)

leakage_features = sorted(
    used_features & forbidden_features
)

print("Number of model features:", len(used_features))
print("Forbidden features used:", leakage_features)

if len(leakage_features) == 0:
    print("Leakage check passed: no forbidden target/ID features are used.")
else:
    print("WARNING: forbidden features detected.")
```

```python
from sklearn.ensemble import RandomForestClassifier

# Random Forest does not need feature scaling
rf_preprocessor = ColumnTransformer([
    ("numeric", SimpleImputer(strategy="median"), MODEL_NUMERIC_FEATURES),
    (
        "categorical",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]),
        MODEL_CATEGORICAL_FEATURES
    )
])

random_forest_model = Pipeline([
    ("preprocessor", rf_preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    ))
])

random_forest_model.fit(X_train, y_train)

rf_scores = random_forest_model.predict_proba(X_test)[:, 1]

print("Random Forest trained successfully.")
print("Number of test predictions:", len(rf_scores))
print("Minimum predicted score:", round(rf_scores.min(), 4))
print("Maximum predicted score:", round(rf_scores.max(), 4))
```

```python
# Compare all methods on the same test set
final_comparison_rows = []

for k in K_VALUES:
    final_comparison_rows.append({
        "Method": "Week-4 Baseline",
        "Metric": f"Precision@{k}",
        "Value": precision_at_k(
            y_test,
            baseline_scores_test,
            k
        )
    })

    final_comparison_rows.append({
        "Method": "Logistic Regression",
        "Metric": f"Precision@{k}",
        "Value": precision_at_k(
            y_test,
            logistic_scores,
            k
        )
    })

    final_comparison_rows.append({
        "Method": "Random Forest",
        "Metric": f"Precision@{k}",
        "Value": precision_at_k(
            y_test,
            rf_scores,
            k
        )
    })

final_comparison_df = pd.DataFrame(final_comparison_rows)

print("Test-set base rate:",
      round(y_test.mean() * 100, 2), "%")

print("\nFinal comparison:")
display(
    final_comparison_df.style.format({
        "Value": "{:.3f}"
    })
)
```

The three approaches were evaluated on the same client-grouped test set using the same Precision@K metric. The test-set base rate of declining content was 51.1%.

At Precision@20, the Week-4 baseline and Logistic Regression both achieved 0.700, while Random Forest achieved 0.750. At Precision@50, Logistic Regression achieved 0.720 compared with 0.680 for Random Forest and 0.620 for the Week-4 baseline. At Precision@100, Logistic Regression and Random Forest both achieved 0.700, compared with 0.550 for the baseline.

The results show that model performance depends on the size of the review set. Random Forest produced the highest measured Precision@20, while Logistic Regression produced higher measured Precision@50 and the same Precision@100 as Random Forest. Therefore, the additional complexity of Random Forest did not provide a consistent improvement across all review depths.

For this task, Precision@K is more useful than accuracy alone because the workflow is intended to prioritize a limited number of content items for review. These results should be treated as directional evidence from this client-grouped test split rather than as proof that one model will perform the same way on future data.
